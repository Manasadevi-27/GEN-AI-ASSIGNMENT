```python
# ============================================================
# WORD EMBEDDING USING ONE-HOT ENCODING AND WORD2VEC
# ============================================================

# ------------------------------------------------------------
# PART 1: WORD EMBEDDING USING ONE-HOT ENCODING
# ------------------------------------------------------------

from sklearn.preprocessing import OneHotEncoder

# List of words
words = [
    ['cat'],
    ['dog'],
    ['apple'],
    ['car']
]

# Create One-Hot Encoder
encoder = OneHotEncoder(sparse_output=False)

# Convert words into one-hot vectors
result = encoder.fit_transform(words)

# Display categories
print("Words:", encoder.categories_[0])

# Display embeddings
print("\nOne-Hot Embeddings:")
print(result)


# ------------------------------------------------------------
# PART 2: FIND SIMILAR WORDS USING WORD2VEC
# ------------------------------------------------------------

# Install gensim if it is not already installed
!pip install -q gensim


# Import Word2Vec
from gensim.models import Word2Vec


# Training sentences
sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"]
]


# Create Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=10,
    window=2,
    min_count=1,
    workers=1
)


# ------------------------------------------------------------
# FIND SIMILAR WORDS
# ------------------------------------------------------------

print("\n========================================")
print("Similar words for 'java'")
print("========================================")

print(model.wv.most_similar("java", topn=3))


print("\n========================================")
print("Similar words for 'python'")
print("========================================")

print(model.wv.most_similar("python", topn=4))


# ------------------------------------------------------------
# DISPLAY WORD VECTORS
# ------------------------------------------------------------

print("\n========================================")
print("Word2Vec Vector for 'python'")
print("========================================")

print(model.wv["python"])


print("\n========================================")
print("Word2Vec Vector for 'java'")
print("========================================")

print(model.wv["java"])
```
